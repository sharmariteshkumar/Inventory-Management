```python
import tkinter as tk
from tkinter import ttk, messagebox


class InventoryApp(tk.Tk):
    def __init__(self):
        super().__init__()

        self.title("Inventory Management System")
        self.geometry("1050x650")
        self.minsize(900, 560)
        self.configure(bg="#f4f6f8")

        self.inventory = {}

        self.configure_style()
        self.build_ui()
        self.refresh_table()

        self.product_entry.focus_set()

    # -----------------------------
    # Styling
    # -----------------------------
    def configure_style(self):
        style = ttk.Style(self)
        style.theme_use("clam")

        style.configure(
            "Treeview",
            background="white",
            foreground="#263238",
            rowheight=38,
            fieldbackground="white",
            font=("Segoe UI", 10),
            borderwidth=0
        )

        style.configure(
            "Treeview.Heading",
            background="#1f4e78",
            foreground="white",
            font=("Segoe UI", 10, "bold"),
            padding=(10, 10)
        )

        style.map(
            "Treeview",
            background=[("selected", "#d9eaf7")],
            foreground=[("selected", "#1f2933")]
        )

    # -----------------------------
    # Build User Interface
    # -----------------------------
    def build_ui(self):

        # Header
        header = tk.Frame(
            self,
            bg="#17365d",
            height=90
        )
        header.pack(fill="x")
        header.pack_propagate(False)

        tk.Label(
            header,
            text="Inventory Management System",
            bg="#17365d",
            fg="white",
            font=("Segoe UI", 22, "bold")
        ).pack(
            anchor="w",
            padx=30,
            pady=(17, 0)
        )

        tk.Label(
            header,
            text="Manage products, quantities, updates and searches",
            bg="#17365d",
            fg="#cfe2f3",
            font=("Segoe UI", 10)
        ).pack(
            anchor="w",
            padx=32
        )

        # Main area
        main = tk.Frame(
            self,
            bg="#f4f6f8"
        )
        main.pack(
            fill="both",
            expand=True,
            padx=25,
            pady=22
        )

        # =============================
        # LEFT PANEL
        # =============================

        form = tk.Frame(
            main,
            bg="white",
            highlightthickness=1,
            highlightbackground="#e1e6eb"
        )

        form.pack(
            side="left",
            fill="y",
            padx=(0, 18)
        )

        form.configure(width=315)
        form.pack_propagate(False)

        tk.Label(
            form,
            text="Product Management",
            bg="white",
            fg="#17365d",
            font=("Segoe UI", 15, "bold")
        ).pack(
            anchor="w",
            padx=22,
            pady=(22, 18)
        )

        # Product Name
        tk.Label(
            form,
            text="Product Name",
            bg="white",
            fg="#4a5568",
            font=("Segoe UI", 9, "bold")
        ).pack(
            anchor="w",
            padx=22
        )

        self.product_entry = ttk.Entry(
            form,
            font=("Segoe UI", 11)
        )

        self.product_entry.pack(
            fill="x",
            padx=22,
            pady=(6, 16),
            ipady=5
        )

        # Quantity
        tk.Label(
            form,
            text="Quantity",
            bg="white",
            fg="#4a5568",
            font=("Segoe UI", 9, "bold")
        ).pack(
            anchor="w",
            padx=22
        )

        self.quantity_entry = ttk.Entry(
            form,
            font=("Segoe UI", 11)
        )

        self.quantity_entry.pack(
            fill="x",
            padx=22,
            pady=(6, 18),
            ipady=5
        )

        # Buttons
        buttons = tk.Frame(
            form,
            bg="white"
        )

        buttons.pack(
            fill="x",
            padx=22
        )

        self.create_button(
            buttons,
            "Add Product",
            self.add_product,
            "#1f7a4d"
        ).pack(
            fill="x",
            pady=4
        )

        self.create_button(
            buttons,
            "Update Quantity",
            self.update_product,
            "#1f4e78"
        ).pack(
            fill="x",
            pady=4
        )

        self.create_button(
            buttons,
            "Delete Product",
            self.delete_product,
            "#b23a48"
        ).pack(
            fill="x",
            pady=4
        )

        self.create_button(
            buttons,
            "Clear Fields",
            self.clear_fields,
            "#6b7280"
        ).pack(
            fill="x",
            pady=4
        )

        # =============================
        # INVENTORY SUMMARY
        # =============================

        stats = tk.Frame(
            form,
            bg="#f7f9fb",
            highlightthickness=1,
            highlightbackground="#e5e9ee"
        )

        stats.pack(
            fill="x",
            padx=22,
            pady=(28, 0)
        )

        tk.Label(
            stats,
            text="Inventory Summary",
            bg="#f7f9fb",
            fg="#17365d",
            font=("Segoe UI", 10, "bold")
        ).pack(
            anchor="w",
            padx=15,
            pady=(12, 4)
        )

        self.product_count_label = tk.Label(
            stats,
            text="Products: 0",
            bg="#f7f9fb",
            fg="#374151",
            font=("Segoe UI", 10)
        )

        self.product_count_label.pack(
            anchor="w",
            padx=15,
            pady=2
        )

        self.total_quantity_label = tk.Label(
            stats,
            text="Total Units: 0",
            bg="#f7f9fb",
            fg="#374151",
            font=("Segoe UI", 10)
        )

        self.total_quantity_label.pack(
            anchor="w",
            padx=15,
            pady=(2, 12)
        )

        # =============================
        # RIGHT PANEL
        # =============================

        content = tk.Frame(
            main,
            bg="white",
            highlightthickness=1,
            highlightbackground="#e1e6eb"
        )

        content.pack(
            side="left",
            fill="both",
            expand=True
        )

        # Top section
        top = tk.Frame(
            content,
            bg="white"
        )

        top.pack(
            fill="x",
            padx=22,
            pady=(20, 12)
        )

        tk.Label(
            top,
            text="Inventory",
            bg="white",
            fg="#17365d",
            font=("Segoe UI", 16, "bold")
        ).pack(
            side="left"
        )

        # Search
        self.search_var = tk.StringVar()

        self.search_var.trace_add(
            "write",
            lambda *args: self.refresh_table()
        )

        search_frame = tk.Frame(
            top,
            bg="#f3f5f7"
        )

        search_frame.pack(
            side="right"
        )

        tk.Label(
            search_frame,
            text="Search",
            bg="#f3f5f7",
            fg="#6b7280",
            font=("Segoe UI", 9)
        ).pack(
            side="left",
            padx=(10, 4)
        )

        ttk.Entry(
            search_frame,
            textvariable=self.search_var,
            width=25,
            font=("Segoe UI", 10)
        ).pack(
            side="left",
            padx=(0, 10),
            ipady=4
        )

        # =============================
        # INVENTORY TABLE
        # =============================

        table_frame = tk.Frame(
            content,
            bg="white"
        )

        table_frame.pack(
            fill="both",
            expand=True,
            padx=22,
            pady=(0, 20)
        )

        columns = (
            "product",
            "quantity",
            "status"
        )

        self.tree = ttk.Treeview(
            table_frame,
            columns=columns,
            show="headings",
            selectmode="browse"
        )

        self.tree.heading(
            "product",
            text="PRODUCT"
        )

        self.tree.heading(
            "quantity",
            text="QUANTITY"
        )

        self.tree.heading(
            "status",
            text="STATUS"
        )

        self.tree.column(
            "product",
            width=260,
            anchor="w"
        )

        self.tree.column(
            "quantity",
            width=120,
            anchor="center"
        )

        self.tree.column(
            "status",
            width=140,
            anchor="center"
        )

        scrollbar = ttk.Scrollbar(
            table_frame,
            orient="vertical",
            command=self.tree.yview
        )

        self.tree.configure(
            yscrollcommand=scrollbar.set
        )

        self.tree.pack(
            side="left",
            fill="both",
            expand=True
        )

        scrollbar.pack(
            side="right",
            fill="y"
        )

        self.tree.bind(
            "<<TreeviewSelect>>",
            self.select_product
        )

        # Footer
        footer = tk.Label(
            self,
            text="Ready • Inventory data is stored during this session",
            bg="#e9edf2",
            fg="#64748b",
            anchor="w",
            font=("Segoe UI", 9),
            padx=25,
            pady=7
        )

        footer.pack(
            fill="x"
        )

    # -----------------------------
    # Create Button
    # -----------------------------
    def create_button(
        self,
        parent,
        text,
        command,
        color
    ):

        button = tk.Button(
            parent,
            text=text,
            command=command,
            bg=color,
            fg="white",
            activebackground=color,
            activeforeground="white",
            relief="flat",
            bd=0,
            cursor="hand2",
            font=("Segoe UI", 10, "bold"),
            padx=10,
            pady=9
        )

        return button

    # -----------------------------
    # Validate Input
    # -----------------------------
    def validate_inputs(self):

        product = self.product_entry.get().strip()
        quantity_text = self.quantity_entry.get().strip()

        if not product:

            messagebox.showwarning(
                "Validation",
                "Please enter a product name."
            )

            self.product_entry.focus_set()

            return None

        try:

            quantity = int(quantity_text)

            if quantity < 0:
                raise ValueError

        except ValueError:

            messagebox.showwarning(
                "Validation",
                "Quantity must be a whole number greater than or equal to 0."
            )

            self.quantity_entry.focus_set()

            return None

        return product, quantity

    # -----------------------------
    # Add Product
    # -----------------------------
    def add_product(self):

        result = self.validate_inputs()

        if not result:
            return

        product, quantity = result

        if product in self.inventory:

            messagebox.showinfo(
                "Product Exists",
                f'"{product}" already exists.\n\n'
                "Use Update Quantity instead."
            )

            return

        self.inventory[product] = quantity

        self.refresh_table()

        self.clear_fields()

        messagebox.showinfo(
            "Success",
            "Product added successfully."
        )

    # -----------------------------
    # Update Product
    # -----------------------------
    def update_product(self):

        result = self.validate_inputs()

        if not result:
            return

        product, quantity = result

        if product not in self.inventory:

            messagebox.showwarning(
                "Product Not Found",
                f'"{product}" does not exist in the inventory.'
            )

            return

        self.inventory[product] = quantity

        self.refresh_table()

        self.clear_fields()

        messagebox.showinfo(
            "Success",
            "Quantity updated successfully."
        )

    # -----------------------------
    # Delete Product
    # -----------------------------
    def delete_product(self):

        product = self.product_entry.get().strip()

        if not product:

            messagebox.showwarning(
                "Validation",
                "Enter a product name to delete."
            )

            return

        if product not in self.inventory:

            messagebox.showwarning(
                "Product Not Found",
                f'"{product}" does not exist in the inventory.'
            )

            return

        confirm = messagebox.askyesno(
            "Confirm Delete",
            f'Delete "{product}" from the inventory?'
        )

        if confirm:

            del self.inventory[product]

            self.refresh_table()

            self.clear_fields()

            messagebox.showinfo(
                "Deleted",
                "Product deleted successfully."
            )

    # -----------------------------
    # Clear Fields
    # -----------------------------
    def clear_fields(self):

        self.product_entry.delete(
            0,
            tk.END
        )

        self.quantity_entry.delete(
            0,
            tk.END
        )

        self.tree.selection_remove(
            self.tree.selection()
        )

        self.product_entry.focus_set()

    # -----------------------------
    # Select Product
    # -----------------------------
    def select_product(self, event=None):

        selection = self.tree.selection()

        if not selection:
            return

        product = self.tree.item(
            selection[0],
            "values"
        )[0]

        self.product_entry.delete(
            0,
            tk.END
        )

        self.product_entry.insert(
            0,
            product
        )

        self.quantity_entry.delete(
            0,
            tk.END
        )

        self.quantity_entry.insert(
            0,
            str(self.inventory[product])
        )

    # -----------------------------
    # Refresh Table
    # -----------------------------
    def refresh_table(self):

        for item in self.tree.get_children():

            self.tree.delete(item)

        query = self.search_var.get().strip().lower()

        for product, quantity in sorted(
            self.inventory.items(),
            key=lambda item: item[0].lower()
        ):

            if query and query not in product.lower():
                continue

            if quantity == 0:

                status = "Out of Stock"

            elif quantity <= 5:

                status = "Low Stock"

            else:

                status = "In Stock"

            self.tree.insert(
                "",
                "end",
                values=(
                    product,
                    quantity,
                    status
                )
            )

        total_units = sum(
            self.inventory.values()
        )

        self.product_count_label.config(
            text=f"Products: {len(self.inventory)}"
        )

        self.total_quantity_label.config(
            text=f"Total Units: {total_units}"
        )


# ==========================================
# START APPLICATION
# ==========================================

if __name__ == "__main__":

    app = InventoryApp()

    app.mainloop()
```
